```python
# Import libraries
import pandas as pd
import numpy as np

# Load the data files
flights_data = pd.read_csv("flights.csv")
weather_data_pd = pd.read_csv("weather.csv")

# Create a NumPy version of the weather data
weather_data_np = weather_data_pd.to_numpy()
```

```python
# Count flights whose origin is JFK and destination is SLC
q_1 = len(flights_data[(flights_data["origin"] == "JFK") &
                       (flights_data["dest"] == "SLC")])

print(q_1)
```

```python
# Count unique airlines/carriers flying to SLC
q_2 = flights_data[flights_data["dest"] == "SLC"]["carrier"].nunique()

print(q_2)
```

```python
# Find the mean arrival delay for flights whose destination is RDU
q_3 = flights_data[flights_data["dest"] == "RDU"]["arr_delay"].mean()

print(q_3)
```

```python
# Select flights going to SEA
sea_flights = flights_data[flights_data["dest"] == "SEA"]

# Calculate the proportion that began at LGA or JFK
q_4 = sea_flights["origin"].isin(["LGA", "JFK"]).mean()

print(q_4)
```

```python
# Create a date column in year/month/day format
flights_data["date"] = (
    flights_data["year"].astype(str) + "/" +
    flights_data["month"].astype(str) + "/" +
    flights_data["day"].astype(str)
)

# Group by date, calculate the average departure delay, and select the largest
q_5 = (
    flights_data.groupby("date", as_index=False)["dep_delay"]
    .mean()
    .sort_values("dep_delay", ascending=False)
    .head(1)
)

print(q_5)
```

```python
# Group by date, calculate the average arrival delay, and select the largest
q_6 = (
    flights_data.groupby("date", as_index=False)["arr_delay"]
    .mean()
    .sort_values("arr_delay", ascending=False)
    .head(1)
)

print(q_6)
```

```python
# Keep flights from JFK or LGA with a known, nonzero air time
nyc_flights = flights_data[
    (flights_data["origin"].isin(["JFK", "LGA"])) &
    (flights_data["air_time"].notna()) &
    (flights_data["air_time"] > 0)
].copy()

# Calculate speed as distance divided by air time
nyc_flights["speed"] = nyc_flights["distance"] / nyc_flights["air_time"]

# Find the row with the highest speed
fastest_index = nyc_flights["speed"].idxmax()

q_7 = nyc_flights.loc[[fastest_index], ["tailnum", "speed"]]

print(q_7)
```

```python
import pandas as pd
import numpy as np

weather_data_pd = pd.read_csv("weather.csv")
weather_data_np = weather_data_pd.to_numpy()
```

```python
# Column 3 contains the month values
month_column = weather_data_np[:, 3].astype(float)

# True for February rows; NaN rows become False
february_mask = month_column == 2

# Count February observations
q_9 = np.sum(february_mask)

print(q_9)
```

```python
# Column 8 contains humidity values
humidity_column = weather_data_np[:, 8].astype(float)

# Keep only humidity values from February
february_humidity = humidity_column[february_mask]

# Calculate the mean, ignoring any missing values
q_10 = np.nanmean(february_humidity)

print(q_10)
```

```python
# Calculate the standard deviation, ignoring any missing values
q_11 = np.nanstd(february_humidity)

print(q_11)
```
